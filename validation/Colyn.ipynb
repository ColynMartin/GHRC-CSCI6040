```python
with open(r"C:\Users\rebec\Downloads\week4_outputs (4)\t_test.txt") as f:
    t_text = f.read()
    print("T-TEST:\n", t_text)

with open(r"C:\Users\rebec\Downloads\week4_outputs (4)\mann_whitney.txt") as f:
    mw_text = f.read()
    print("\nMANN-WHITNEY:\n", mw_text)

with open(r"C:\Users\rebec\Downloads\week4_outputs (4)\cohens_d.txt") as f:
    d_text = f.read()
    print("\nCOHENS D:\n", d_text)

with open(r"C:\Users\rebec\Downloads\week4_outputs (4)\threshold.txt") as f:
    th_text = f.read()
    print("\nTHRESHOLD:\n", th_text)
```

```python
import re

# -----------------------------
# Extract values
# -----------------------------
def get_number(text, label):
    match = re.search(label + r".*?(-?[0-9]+\.?[0-9]*(?:[eE][+-]?[0-9]+)?)", text, re.IGNORECASE)
    return float(match.group(1)) if match else None

t_p = get_number(t_text, "p")
mw_p = get_number(mw_text, "p")
d_val = get_number(d_text, "d")
threshold = get_number(th_text, "threshold")

# -----------------------------
# Interpret results
# -----------------------------
print("\n===== INTERPRETATION =====")

# P-values
if t_p is not None:
    print(f"T-test p-value: {t_p} →", "Significant" if t_p < 0.05 else "Not significant")

if mw_p is not None:
    print(f"Mann-Whitney p-value: {mw_p} →", "Significant" if mw_p < 0.05 else "Not significant")

# Effect size
if d_val is not None:
    if abs(d_val) >= 0.8:
        effect = "Large"
    elif abs(d_val) >= 0.5:
        effect = "Medium"
    else:
        effect = "Small"
    print(f"Cohen's d: {d_val} → {effect} effect")

# Threshold
if threshold is not None:
    print(f"Threshold used: {threshold}")
```
